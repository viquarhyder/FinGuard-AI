#### Core Imports

```python
import os
import sys
import warnings
from pathlib import Path

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")

print("FinGuard AI — Data Loading")
print("=" * 60)
print(f"Python Version : {sys.version.split()[0]}")
print(f"NumPy Version  : {np.__version__}")
print(f"Pandas Version : {pd.__version__}")
```

#### Project Paths

```python
# ============================================================
# Project Directory Configuration
# ============================================================

PROJECT_ROOT = Path.cwd().parent

DATASET_DIR = PROJECT_ROOT / "Dataset"
RAW_DATA_DIR = DATASET_DIR / "Raw"
PROCESSED_DATA_DIR = DATASET_DIR / "Processed"
SAMPLE_DATA_DIR = DATASET_DIR / "Sample"

REPORTS_DIR = PROJECT_ROOT / "Reports"
FIGURES_DIR = REPORTS_DIR / "Figures"

RAW_DATA_PATH = RAW_DATA_DIR / "creditcard.csv"

print("Project Paths")
print("=" * 60)

print(f"Project Root       : {PROJECT_ROOT}")
print(f"Dataset Directory  : {DATASET_DIR}")
print(f"Raw Dataset        : {RAW_DATA_DIR}")
print(f"Processed Dataset  : {PROCESSED_DATA_DIR}")
print(f"Sample Dataset     : {SAMPLE_DATA_DIR}")
print(f"Raw File           : {RAW_DATA_PATH}")
```

#### Validate Project Structure

```python
# ============================================================
# Project Structure Validation
# ============================================================

required_directories = [
    DATASET_DIR,
    RAW_DATA_DIR,
    PROCESSED_DATA_DIR,
    SAMPLE_DATA_DIR,
    PROJECT_ROOT / "Models",
    PROJECT_ROOT / "src",
    PROJECT_ROOT / "templates",
    PROJECT_ROOT / "static",
    PROJECT_ROOT / "Reports",
    PROJECT_ROOT / "Tests",
    PROJECT_ROOT / "config",
    PROJECT_ROOT / "api",
]

print("Project Structure Validation")
print("=" * 60)

all_directories_exist = True

for directory in required_directories:
    exists = directory.exists()
    
    if exists:
        status = "✓ EXISTS"
    else:
        status = "✗ MISSING"
        all_directories_exist = False
    
    print(f"{status:<12} {directory.relative_to(PROJECT_ROOT)}")

print("=" * 60)

if all_directories_exist:
    print("✓ Project structure validation PASSED")
else:
    print("⚠ Some project directories are missing")
```

#### Validate Raw Dataset

```python
# ============================================================
# Raw Dataset Validation
# ============================================================

print("Raw Dataset Validation")
print("=" * 60)

if RAW_DATA_PATH.exists():
    file_size_mb = RAW_DATA_PATH.stat().st_size / (1024 ** 2)
    
    print("✓ Dataset file found")
    print(f"File Name : {RAW_DATA_PATH.name}")
    print(f"File Size : {file_size_mb:.2f} MB")
    print(f"Location  : {RAW_DATA_PATH}")
else:
    raise FileNotFoundError(
        f"Raw dataset not found at:\n{RAW_DATA_PATH}"
    )
```

#### 

#### Discover Raw Files

```python
# ============================================================
# Raw Dataset Directory Inspection
# ============================================================

raw_files = [
    file for file in RAW_DATA_DIR.iterdir()
    if file.is_file()
]

print("Files Found in Dataset/Raw")
print("=" * 60)

for file in raw_files:
    size_mb = file.stat().st_size / (1024 ** 2)
    print(f"{file.name:<30} {size_mb:>10.2f} MB")

print("=" * 60)
print(f"Total files: {len(raw_files)}")
```

#### Inspect Dataset Header

```python
# ============================================================
# Dataset Header Inspection
# ============================================================

header_df = pd.read_csv(
    RAW_DATA_PATH,
    nrows=5
)

print("Dataset Columns")
print("=" * 60)

for index, column in enumerate(header_df.columns, start=1):
    print(f"{index:02d}. {column}")

print("=" * 60)
print(f"Total Columns: {len(header_df.columns)}")
```

#### Preview Raw Data

```python
# ============================================================
# Raw Dataset Preview
# ============================================================

print("First 5 Transactions")
print("=" * 60)

display(header_df)
```

#### Dataset Shape

```python
# ============================================================
# Dataset Row Count
# ============================================================

print("Counting Dataset Rows")
print("=" * 60)

row_count = 0

for chunk in pd.read_csv(
    RAW_DATA_PATH,
    usecols=["Class"],
    chunksize=50_000
):
    row_count += len(chunk)

print(f"Total Transactions : {row_count:,}")
print(f"Total Features     : {len(header_df.columns) - 1}")
print(f"Target Column      : Class")
```

#### Load Full Dataset Efficiently

```python
# ============================================================
# Full Dataset Loading
# ============================================================

print("Loading Full Dataset...")
print("=" * 60)

# Memory-efficient dtypes
dtype_map = {
    "Time": "float32",
    "Amount": "float32",
    "Class": "int8"
}

# PCA-transformed features
for column in header_df.columns:
    if column.startswith("V"):
        dtype_map[column] = "float32"

df = pd.read_csv(
    RAW_DATA_PATH,
    dtype=dtype_map
)

print("✓ Dataset loaded successfully")
print("=" * 60)
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

#### Dataset Shape Verification

```python
# ============================================================
# Shape Verification
# ============================================================

EXPECTED_ROWS = 284_807
EXPECTED_COLUMNS = 31

print("Dataset Shape Verification")
print("=" * 60)

print(f"Actual Rows       : {df.shape[0]:,}")
print(f"Expected Rows     : {EXPECTED_ROWS:,}")

print(f"Actual Columns    : {df.shape[1]}")
print(f"Expected Columns  : {EXPECTED_COLUMNS}")

if df.shape == (EXPECTED_ROWS, EXPECTED_COLUMNS):
    print("\n✓ Dataset shape validation PASSED")
else:
    print("\n⚠ Dataset shape differs from expected dimensions")
```

#### Data Types

```python
# ============================================================
# Data Type Inspection
# ============================================================

print("Data Types")
print("=" * 60)

display(
    pd.DataFrame({
        "Column": df.columns,
        "Data_Type": df.dtypes.astype(str).values
    })
)
```

#### Missing Values

```python
# ============================================================
# Missing Value Analysis
# ============================================================

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": missing_values.index,
    "Missing_Count": missing_values.values,
})

missing_report["Missing_Percentage"] = (
    missing_report["Missing_Count"]
    / len(df)
    * 100
)

missing_report = missing_report.sort_values(
    by="Missing_Count",
    ascending=False
)

print("Missing Value Analysis")
print("=" * 60)

display(missing_report)

if missing_values.sum() == 0:
    print("✓ No missing values detected")
else:
    print("⚠ Missing values detected")
```

#### Duplicate Transactions

```python
# ============================================================
# Duplicate Transaction Analysis
# ============================================================

duplicate_count = df.duplicated().sum()

print("Duplicate Transaction Analysis")
print("=" * 60)
print(f"Duplicate Rows : {duplicate_count:,}")

if duplicate_count == 0:
    print("✓ No duplicate rows detected")
else:
    print("⚠ Duplicate rows detected")
```

#### Target Validation

```python
# ============================================================
# Target Variable Validation
# ============================================================

TARGET_COLUMN = "Class"

unique_target_values = sorted(
    df[TARGET_COLUMN].unique().tolist()
)

print("Target Variable Validation")
print("=" * 60)

print(f"Target Column : {TARGET_COLUMN}")
print(f"Unique Values : {unique_target_values}")

expected_classes = {0, 1}

if set(unique_target_values) == expected_classes:
    print("✓ Target values validation PASSED")
else:
    print("⚠ Unexpected target values detected")
```

#### Class Distribution

```python
# ============================================================
# Fraud Class Distribution
# ============================================================

class_counts = df[TARGET_COLUMN].value_counts().sort_index()

class_distribution = pd.DataFrame({
    "Class": class_counts.index,
    "Transaction_Count": class_counts.values
})

class_distribution["Percentage"] = (
    class_distribution["Transaction_Count"]
    / len(df)
    * 100
)

class_distribution["Label"] = class_distribution["Class"].map({
    0: "Legitimate",
    1: "Fraud"
})

print("Transaction Class Distribution")
print("=" * 60)

display(class_distribution)
```

#### Fraud Statistics

```python
# ============================================================
# Fraud Statistics
# ============================================================

legitimate_count = int((df["Class"] == 0).sum())
fraud_count = int((df["Class"] == 1).sum())

fraud_rate = fraud_count / len(df) * 100

print("Fraud Statistics")
print("=" * 60)

print(f"Total Transactions : {len(df):,}")
print(f"Legitimate        : {legitimate_count:,}")
print(f"Fraudulent        : {fraud_count:,}")
print(f"Fraud Rate        : {fraud_rate:.4f}%")
print(f"Legitimate Rate    : {100 - fraud_rate:.4f}%")
```

#### Amount Validation

```python
# ============================================================
# Transaction Amount Validation
# ============================================================

amount_statistics = df["Amount"].describe()

print("Transaction Amount Statistics")
print("=" * 60)

display(amount_statistics.to_frame(name="Amount"))
```

#### Time Validation

```python
# ============================================================
# Transaction Time Validation
# ============================================================

time_statistics = df["Time"].describe()

print("Transaction Time Statistics")
print("=" * 60)

display(time_statistics.to_frame(name="Time"))
```

#### Memory Usage

```python
# ============================================================
# Memory Usage Analysis
# ============================================================

memory_usage_mb = (
    df.memory_usage(deep=True).sum()
    / (1024 ** 2)
)

print("Memory Usage")
print("=" * 60)
print(f"DataFrame Memory : {memory_usage_mb:.2f} MB")
```

#### Feature Summary

```python
# ============================================================
# Feature Summary
# ============================================================

feature_summary = pd.DataFrame({
    "Feature": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

print("Feature Summary")
print("=" * 60)

display(feature_summary)
```

#### Create Representative Sample

```python
# ============================================================
# Representative Sample Dataset
# ============================================================

SAMPLE_PATH = SAMPLE_DATA_DIR / "sample_transactions.csv"

fraud_sample = df[df["Class"] == 1]

legitimate_sample = df[df["Class"] == 0].sample(
    n=min(9_508, legitimate_count),
    random_state=42
)

sample_df = pd.concat(
    [fraud_sample, legitimate_sample],
    axis=0
).sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

sample_df.to_csv(
    SAMPLE_PATH,
    index=False
)

print("Representative Sample Created")
print("=" * 60)
print(f"Sample Path : {SAMPLE_PATH}")
print(f"Sample Rows : {len(sample_df):,}")
print(f"Fraud Rows  : {(sample_df['Class'] == 1).sum():,}")
print(f"Legit Rows  : {(sample_df['Class'] == 0).sum():,}")
```

#### Sample Verification

```python
# ============================================================
# Sample Dataset Verification
# ============================================================

sample_check = pd.read_csv(SAMPLE_PATH)

print("Sample Dataset Verification")
print("=" * 60)

print(f"Rows    : {sample_check.shape[0]:,}")
print(f"Columns : {sample_check.shape[1]}")
print(f"Fraud   : {(sample_check['Class'] == 1).sum():,}")
print(f"Legit   : {(sample_check['Class'] == 0).sum():,}")

display(sample_check.head())
```

#### Data Loading Summary

```python
# ============================================================
# Final Data Loading Summary
# ============================================================

print("=" * 70)
print("FIN GUARD AI — DATA LOADING SUMMARY")
print("=" * 70)

print(f"✓ Raw Dataset       : {RAW_DATA_PATH.name}")
print(f"✓ Total Transactions: {len(df):,}")
print(f"✓ Total Columns     : {df.shape[1]}")
print(f"✓ Fraud Cases       : {fraud_count:,}")
print(f"✓ Fraud Rate        : {fraud_rate:.4f}%")
print(f"✓ Missing Values    : {int(df.isnull().sum().sum()):,}")
print(f"✓ Duplicate Rows    : {duplicate_count:,}")
print(f"✓ Memory Usage      : {memory_usage_mb:.2f} MB")
print(f"✓ Sample Dataset    : {SAMPLE_PATH.name}")

print("=" * 70)
print("✓ NOTEBOOK 01 DATA LOADING COMPLETED")
print("=" * 70)
```
